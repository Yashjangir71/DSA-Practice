```python
from gensim.models import Word2Vec 
```

```python
corpus = [    "machine learning is fun",    "deep learning is a part of machine learning",    "word embeddings capture meaning of words",    "neural networks learn useful representations" ] 
```

```python
sentences = [s.lower().split() for s in corpus] 
model = Word2Vec(sentences=sentences, vector_size=50, window=3, min_count=1, sg=1, epochs=100) 

print("Training complete.") 
print("Embedding for 'networks':") 
print(model.wv["networks"])
```

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D
from tensorflow.keras.layers import Flatten, Dense

(x_train, y_train), (x_test, y_test) = mnist.load_data()

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

x_train = x_train[..., np.newaxis]
x_test = x_test[..., np.newaxis]

model = Sequential([
    Conv2D(32, (3, 3), activation="relu",
           input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(64, activation="relu"),
    Dense(10, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

model.fit(
    x_train, y_train,
    epochs=3,
    batch_size=128,
    validation_split=0.1
)

loss, accuracy = model.evaluate(x_test, y_test)

print("Test Accuracy:", accuracy)

prediction = model.predict(x_test[:1])
print("Predicted Digit:", np.argmax(prediction[0]))
print("Actual Digit:", y_test[0])

plt.imshow(x_test[0].squeeze(), cmap="gray")
plt.title("MNIST Handwritten Digit")
plt.show()
```
